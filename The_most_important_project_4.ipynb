```python
total_Empolyee=df['EmployeeCount'].count()
print(total_Empolyee)
```

```python
left_Empolyee=len(df[df['Attrition'] == 'Yes'])
print(left_Empolyee)
```

```python
p_loss = left_Empolyee / total_Empolyee
print(p_loss)
```

```python
i=len(df[df['Attrition']=='No'])
print(i)
```

```python
h=i/total_Empolyee
print(h)
```

```python
l=h-p_loss
print(l)
```

```python
sales_now=len(df[(df['Department']=='Sales') & (df['Attrition']=='Yes')])
walk=[sales_now]
p_agian=0.05
p_stay = 1 - (p_loss + p_agian)
for i in range(365):
  change=np.random.choice([1,-1,0],p=[p_loss,p_agian,p_stay])
  new_value = walk[-1] + change
  walk.append(new_value)
plt.figure(figsize=(10, 5))
plt.plot(walk, label='Sales Staff Forecast', color='green')
plt.axhline(y=sales_now, color='red', linestyle='--', label='Start Point')
plt.title('How Sales Staff Count Might Change Over 1 Year')
plt.xlabel('Days')
plt.ylabel('Number of Employees')
plt.legend()
plt.grid(True)
plt.show()#<><>>>>>>>>>>>>>>>>><><><><><><><><><><.
```

```python
df=pd.get_dummies(df,columns=['Attrition','Gender','OverTime'],drop_first=True)
```

```python
df.columns
```

```python
le=LabelEncoder()
df['BusinessTravel']=le.fit_transform(df['BusinessTravel'])
df['Department']=le.fit_transform(df['Department'])
df['EducationField']=le.fit_transform(df['EducationField'])
df['JobRole']=le.fit_transform(df['JobRole'])
df['MaritalStatus']=le.fit_transform(df['MaritalStatus'])
df['Over18']=le.fit_transform(df['Over18'])
```

```python
X=df.drop('Attrition_Yes',axis=1)
Y=df['Attrition_Yes']
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
rf=RandomForestClassifier(n_estimators=100,class_weight='balanced',max_depth=10,random_state=42)
rf.fit(X_train,Y_train)
```

```python
y_pred=rf.predict(X_test)
```

```python
results = pd.DataFrame({
    'Actual (Y_test)': Y_test,
    'Predicted (y_pred)': y_pred
})

# نعرض أول 10 صفوف بس عشان نشوفهم بوضوح
print(results.head(10))
```

```python
print(classification_report(Y_test,y_pred))
```

```python
co=confusion_matrix(Y_test,y_pred)
plt.figure(figsize=(8,5))
sns.heatmap(co,annot=True,fmt='d',cmap='pink')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```
